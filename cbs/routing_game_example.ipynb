```python
from mapf import *
from routing_game import *
from rcbs import random_problem

N_grid = 512
N_region = 32
env = GridWorld((N_grid,N_grid), [])
parts = partition_gridworld(env, (N_region,N_region))
net = SimpleRoutingNet(env, parts)
N_agents = 1024
start, goals = random_problem(N_agents, env)
start_times = N_agents*[0]
prob = MAPFProblem(list(start.keys()), start_times, start, goals)
```

```python
congestion_model = AffineCongestion(
    {frozenset(edge): .5 for edge in net.edges},
    {frozenset(edge): N_region for edge in net.edges}
)
demands = make_demands(net, prob)
routing_game = RoutingGame(net, congestion_model, demands)
```

```python
from fw import *
stuff = run_frank_wolfe(routing_game, ftol_abs = 1, ftol_rel = 1e-4)
```
